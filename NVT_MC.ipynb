```python
import os, shutil, math, glob

parent_dir = '/storage/home/hcoda1/5/sjamdade3/scratch/August_13_2023/water_screening/MD_MC/Manuscript_Final/Defects/'

path_general_files = '/storage/home/hcoda1/5/sjamdade3/scratch/August_13_2023/water_screening/MD_MC/Manuscript_Final/Defects/SOP_files/'

MOF_type = 'defects/'
```

```python
##### NVT MC 1 cycle 

path_CIFs = os.path.join(parent_dir, 'MOFs', str(MOF_type))

path_MOF_NVT_MC = os.path.join(path_CIFs, 'NVT_MC/')

os.mkdir(path_MOF_NVT_MC)

path_NVT_MC_input_files = os.path.join(path_general_files, 'NVT_MC_input_files/')

path_python_files = os.path.join(path_general_files, 'python_files/')


path_MOF_VF = os.path.join(path_CIFs, 'VF/')

files_CIFs = os.listdir(path_CIFs)

files_input = os.listdir(path_NVT_MC_input_files)

index = 0

for f1 in glob.glob(path_CIFs+"*.cif"):
        index  = index  + 1
        
        partitioned_string = f1.partition(str(MOF_type))
        f1 = partitioned_string[-1]

        path = os.path.join(path_MOF_NVT_MC, str(index))
        os.mkdir(path)
        shutil.copy(path_CIFs+f1, path)
        for f2 in files_input:
            shutil.copy(path_NVT_MC_input_files+f2, path)
        f = open(path+'/'+f1,'r')
        lines=f.readlines()
        f.close()
        for line in lines:
            row=line.split()

            if not line.strip():
                continue

            if row[0] == '_cell_length_a':
                a = float(row[1])
            if row[0] == '_cell_length_b':
                b = float(row[1])
            if row[0] == '_cell_length_c':
                c = float(row[1])
            if row[0] == '_cell_angle_alpha':
                alpha = float(row[1])
            if row[0] == '_cell_angle_beta':
                beta = float(row[1])
            if row[0] == '_cell_angle_gamma':
                gamma = float(row[1])

        AX = int(max(math.ceil(20.0/float(a*math.sin(gamma*3.14/180))),math.ceil(20.0/float(b*math.sin(gamma*3.14/180)))))
        BY = int(max(math.ceil(20.0/float(b*math.sin(alpha*3.14/180))),math.ceil(20.0/float(c*math.sin(alpha*3.14/180)))))
        CZ = int(max(math.ceil(20.0/float(c*math.sin(beta*3.14/180))),math.ceil(20.0/float(a*math.sin(beta*3.14/180)))))
       	
        fin = open(''+str(path)+'/simulation.input', "rt")

        data1 = fin.read()
        data1 = data1.replace('AX', str(AX))
        data1 = data1.replace('BY', str(BY))
        data1 = data1.replace('CZ', str(CZ))

        MOF_Name = str((os.path.splitext(str(f1))[0]))

        data1 = data1.replace('MOF', str(MOF_Name))
        
        path_VF_output = os.path.join(path_MOF_VF, str(index), 'Output', 'System_0')
        
        for f_x in os.listdir(path_VF_output):
            if f_x.endswith('.data'):
                print(str(f_x))
                path_file_x = os.path.join(path_VF_output, str(f_x))
                file_x =  open( path_file_x,'r')
                lines=file_x.readlines()
                file_x.close()
                for line in lines:
                    row=line.split()

                    if not line.strip():
                        continue

                    if len(row)==6 and row[0] == 'volume' and row[1] == 'of' and row[2] == 'the' and row[3] == 'cell:':
                        vol_cell = float(row[4])
                        
                    if len(row)==8 and row[0] == '[helium]' and row[1] == 'Average' and row[2] == 'Widom' and row[3] == 'Rosenbluth-weight:':
                        void_fr = float(row[4])
                        
        No_of_molecules = int(1000*vol_cell*(10**-30)*void_fr*1000*6.02214*(10**23)/18)
        
        data1 = data1.replace('Saturated_loading', str(int(No_of_molecules)))

        fin.close()

        fin = open(''+str(path)+'/simulation.input', "wt")

        fin.write(data1)

        fin.close()
```
